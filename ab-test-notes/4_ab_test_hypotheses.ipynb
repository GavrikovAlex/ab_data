```python
import pandas as pd
import numpy as np
```

```python
# Будем проверять честность монетки
# Иммитируем бросок монетки
# Первый аргумент - кол-во бросков, второй аргумент - вероятность выпадения орла
np.random.binomial(10, 0.5) # с вероятностью 50% выпадает орел , т.е. 1
```

Повторяем броски 1000 раз. И дальше считаем отклонение от 5. Потом считает процент ситуаций, в которых это отклонение >=4

# FPR

```python
n = 100000
result = []

for i in range(n):
  np.random.seed(i) # делаем эксперимент воспроизовдимым (чтобы числа оставались одинаковыми)
  result.append(np.random.binomial(10, 0.5))
```

```python
df = pd.DataFrame(result)
```

```python
df
```

```python
# считаем отклонение
df['deviation'] = abs(5 - df[0])
```

```python
df
```

```python
# отклонение
(df['deviation'] >= 4).mean()
```

# Coin Check (проверка монеты)

MDE: 10%

Power (1 - beta) : 80% - 1 - вероятность ошибки II рода, вероятность ложного пропуска

Significance (alpha): 5% - вероятность ошибки I рода

## FPR

```python
sample_size = 200
```

```python
n = 100000
result = []
for i in range(n):
  np.random.seed(i)
  result.append(np.random.binomial(sample_size,0.5))
```

```python
df = pd.DataFrame(result)
```

```python
df
```

```python
df['deviation'] = abs(sample_size * 0.5 - df[0])
```

```python
# пророг
threshold = sample_size * 0.07
threshold
```

```python
(df['deviation'] >= threshold).mean()
```
